```python
import numpy as np
import pandas as pd
import datetime
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colors
import seaborn as sns
import duckdb
```

```python
df = pd.read_excel('urbox_bi_analyst_test.xlsx', sheet_name = 'Data for Part 1')
```

```python
df.head()
```

```python
df.info()
```

```python
query = """ select t1.brand_id brand_1, t2.brand_id brand_2, count(t1.user_id) no_users
from df t1
left join df t2 on t1.user_id = t2.user_id and t1.brand_id != t2.brand_id
where t2.brand_id is not null
group by brand_1, brand_2
having count(t1.user_id) > 3
order by no_users desc
"""
```

```python
result = duckdb.query(query).to_df()
print(result.sort_values(by= 'no_users', ascending = False))
```
